```python
import xml.etree.ElementTree as ET
import re

def extract_process_data(xml_file, output_file):
    tree = ET.parse(xml_file)
    root = tree.getroot()

    # Remove the big, irrelevant sections
    for tag in ['UserVarDatabase', 'CustomUnitSets', 'Basis', 'CaseProperties', 'PFDs']:
        for elem in root.findall(tag):
            root.remove(elem)

    # Write the stripped XML (still readable, but large)
    tree.write(output_file, encoding='utf-8', xml_declaration=True)

    # Compress whitespace to shrink further (optional but recommended)
    with open(output_file, 'r', encoding='utf-8') as f:
        content = f.read()
    compressed = re.sub(r'>\s+<', '><', content)
    with open(output_file, 'w', encoding='utf-8') as f:
        f.write(compressed)

    print(f" Stripped XML saved to: {output_file}")

# Run it
extract_process_data('./Comp.XML', './miniComp.XML')
```
